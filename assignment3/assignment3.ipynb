```python
sql = '''
SELECT "DISTRICT",
      "Lat",
      "Long"
FROM "b973d8cb-eeb2-4e7e-99da-c92938efc9c0"
    '''
```

```python
bpd_logs = "https://data.boston.gov/api/3/action/datastore_search_sql"
```

```python
import urllib, requests
from collections import Counter
```

```python
url = bpd_logs+"?sql="+urllib.parse.quote(sql)
rows = requests.get( url ).json()

data = rows['result']['records']
```

```python
Lat = []
Long = []
District = []
for item in data:
  District.append(item["DISTRICT"])
  if item["Lat"] == None:
    continue
  else:
    Lat.append(item["Lat"])
    Long.append(item["Long"])

item_count = Counter(District)
```
